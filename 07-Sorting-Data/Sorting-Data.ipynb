## Sorting Data
En este notebook exploraremos cómo ordenar datos en Pandas. El ordenamiento es esencial para comprensión, análisis y visualización de datos.

### Plan de aprendizaje:
- Ordenar basándose en una columna
- Ordenar basándose en múltiples columnas
- Ordenar basándose en índices
- Métodos adicionales: nlargest() y nsmallest()

```python
import pandas as pd
import numpy as np
```

## Ordenando por una Columna
Usar sort_values() para ordenar datos en orden ascendente o descendente.

```python
# Crear un DataFrame de ejemplo
df = pd.DataFrame({
    'Name': ['John', 'Jane', 'Bob', 'Alice'],
    'Age': [25, 30, 22, 28],
    'Salary': [50000, 60000, 45000, 55000]
})
df
```

```python
# Ordenar por columna Country en orden ascendente
df.sort_values(by='Age')
```

```python
# Ordenar por columna Country en orden descendente
df.sort_values(by='Age', ascending=False)
```

## Ordenando por Múltiples Columnas
Pasar una lista de columnas a sort_values() para ordenar por varias columnas.

```python
# Ordenar por múltiples columnas (Age ascendente, Salary descendente)
df.sort_values(by=['Age', 'Salary'], ascending=[True, False])
```

## Ordenando por Índice
Usar sort_index() para ordenar los índices (filas o columnas).

```python
# Ordenar por índice en orden ascendente
df.sort_index()
```

```python
# Ordenar por índice en orden descendente
df.sort_index(ascending=False)
```

```python
# Ordenar nombres de columnas en orden alfabético
df.sort_index(axis='columns')
```

## Métodos Adicionales: nlargest() y nsmallest()
Obtener los n valores más grandes o más pequeños de una columna.

```python
# Obtener las 2 filas con los salarios más altos
df.nlargest(2, 'Salary')
```

```python
# Obtener los 2 valores más grandes de la columna Salary
df['Salary'].nlargest(2)
```

```python
# Obtener las 2 filas con las edades más bajas
df.nsmallest(2, 'Age')
```

```python
# Obtener los 2 valores más pequeños de la columna Age
df['Age'].nsmallest(2)
```

## Conclusión
El ordenamiento es fundamental en análisis de datos. Permite identificar patrones, encontrar extremos y preparar datos para visualización e insights significativos.
