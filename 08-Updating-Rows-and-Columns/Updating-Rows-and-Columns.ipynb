## Updating Rows and Columns
En este notebook exploraremos cómo actualizar filas y columnas en DataFrames. Aprender a modificar datos es esencial para la limpieza de datos.

### Plan de aprendizaje:
- Actualizar nombres de columnas
- Actualizar filas completas
- Actualizar filas específicas basándose en condiciones
- Usar apply, map y replace para transformaciones

```python
import pandas as pd
import numpy as np
```

## Actualizando Nombres de Columnas
Usar rename() para cambiar nombres de columnas de forma permanente.

```python
# Crear un DataFrame de ejemplo
Student = {
    "Id": ['cs-01', 'mth-02', 'eng-03', 'phy-04', 'mth-05', 'cs-06'],
    "Name": ["Fatima Khan", 'Noureen Ali', 'Salman Aslam', 'Sana Khan', 'Adam Jack', 'Ali Khan'],
    "subject": ['computer science', 'math', 'english', 'physics', 'statistics', 'programming'],
    "marks": [45, 90, 60, 37, 49, 50]
}
data = pd.DataFrame(Student)
data
```

```python
# Renombrar una columna específica
data.rename(columns={'Name': 'Full Name'}, inplace=True)
data
```

```python
# Convertir todos los nombres de columnas a minúsculas
data.columns = data.columns.str.lower()
data
```

## Actualizando Filas
Usar loc[] para actualizar filas completas o valores específicos.

```python
# Seleccionar una fila completa
data.loc[4]
```

```python
# Actualizar una fila completa
data.loc[4] = ['cs-05', 'Ali', 'Database', 47]
data.loc[4]
```

```python
# Seleccionar columnas específicas de una fila
data.loc[0, ['name', 'subject']]
```

```python
# Actualizar solo columnas específicas
data.loc[0, ['name', 'subject']] = ['Farhan', 'Python']
data.loc[0, ['name', 'subject']]
```

## Actualizando Filas Basándose en Condiciones
Usar condiciones booleanas con loc[] para actualizar múltiples filas.

```python
# Crear una condición: notas menores a 50
filt = data['marks'] < 50
filt
```

```python
# Actualizar filas que cumplen la condición
data.loc[filt, 'marks'] = 'Fail'
data
```

## Conclusión
Actualizar filas y columnas es esencial para la limpieza y preparación de datos. Esto permite transformar datos crudos en información analizable.
