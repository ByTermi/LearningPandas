## Handling Missing Values
En este notebook exploraremos cómo detectar, eliminar y rellenar valores faltantes. Los valores NaN son comunes en datos reales y deben manejarse correctamente.

### Plan de aprendizaje:
- Detectar valores nulos (isnull, notnull)
- Eliminar valores nulos (dropna)
- Lidiar con valores faltantes personalizados
- Rellenar datos faltantes (fillna)

```python
import pandas as pd
import numpy as np
```

## Detectando Valores Nulos
Usar isnull() para identificar valores faltantes (NaN, None).

```python
# Crear un DataFrame con valores faltantes y valores textuales que parecen nulos
student = {
    'first': ['Fatima', 'Carine', 'Alla', 'Salman', np.nan, None, 'NA'],
    'last': ['Khan', 'Aries', 'Ben', 'Malik', np.nan, np.nan, 'NaN'],
    'email': ['fatimakhan@gmail.com', 'caries@email.com', 'allaben@gmail.com', None, np.nan, 'saqib@gmail.com', 'NA'],
    'age': [33, 55, 63, 36, np.nan, None, '?']
}
data = pd.DataFrame(student)
data
```

```python
# Detectar valores nulos con isnull()
data.isnull()
```

```python
# Contar valores nulos por columna
data.isnull().sum()
```

## Eliminando Valores Nulos
Usar dropna() para eliminar filas o columnas con valores faltantes.

```python
# Eliminar filas con AL MENOS un valor nulo (how='any')
data.dropna(how='any')
```

```python
# Eliminar columnas con AL MENOS un valor nulo (axis=1)
data.dropna(how='any', axis=1)
```

```python
# Eliminar columnas solo si TODAS tienen valores nulos (how='all', axis=1)
data.dropna(how='all', axis=1)
```

## Manejando Valores Faltantes Personalizados
Convertir valores textuales que parecen nulos (como 'NA', 'NaN', '?') en NaN verdaderos.

```python
# Reemplazar valores textuales que parecen nulos con NaN verdaderos
data.replace({'NA': np.nan, 'NaN': np.nan, '?': np.nan}, inplace=True)
data
```

```python
# Ahora eliminar todas las filas con valores nulos (incluyendo los que acabamos de convertir)
data.dropna()
```

## Rellenando Valores Faltantes
Usar fillna() para llenar valores nulos con un valor específico o estadístico.

```python
# Recrear el DataFrame original para llenar en lugar de eliminar
student = {
    'first': ['Fatima', 'Carine', 'Alla', 'Salman', np.nan, None, 'NA'],
    'last': ['Khan', 'Aries', 'Ben', 'Malik', np.nan, np.nan, 'NaN'],
    'email': ['fatimakhan@gmail.com', 'caries@email.com', 'allaben@gmail.com', None, np.nan, 'saqib@gmail.com', 'NA'],
    'age': [33, 55, 63, 36, np.nan, None, '?']
}
data = pd.DataFrame(student)
data.replace({'NA': np.nan, 'NaN': np.nan, '?': np.nan}, inplace=True)
data
```

```python
# Llenar todos los NaN con un valor específico
data.fillna('MISSING')
```

```python
# Llenar NaN en la columna 'age' con la media (solo para columnas numéricas)
# Primero convertir 'age' a numérico
data_numeric = data.copy()
data_numeric['age'] = pd.to_numeric(data_numeric['age'], errors='coerce')
data_numeric['age'].fillna(data_numeric['age'].mean(), inplace=True)
data_numeric
```

## Conclusión
El manejo de valores faltantes es crítico en ciencia de datos. Elegir entre eliminar o rellenar datos afecta directamente la calidad del análisis y el rendimiento de modelos de ML.
