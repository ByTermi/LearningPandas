## Adding and Dropping Rows and Columns
En este notebook exploraremos cómo agregar y eliminar filas y columnas, así como combinar datasets. Estos son conceptos clave para la preparación de datos.

### Plan de aprendizaje:
- Agregar columnas con operador de indexación
- Agregar columnas con método Insert
- Eliminar columnas
- Agregar filas con método Concat
- Eliminar filas

```python
import pandas as pd
```

## Agregando Columnas con Operador de Indexación
Crear nuevas columnas asignando cálculos a un nombre de columna inexistente.

```python
# Crear un DataFrame de ejemplo
df = pd.DataFrame({
    'Category': ['Breakfast', 'Breakfast', 'Lunch', 'Lunch', 'Dinner'],
    'Item': ['Eggs', 'Toast', 'Burger', 'Salad', 'Pasta'],
    'Calories': [155, 100, 354, 150, 220],
    'Carbohydrates': [1.1, 14, 36, 4, 43],
    'Protein': [13, 4, 12, 8, 9]
})
df
```

```python
# Agregar una columna calculada: Carbohydrates Percentage
df['Carbohydrates Percentage'] = df['Carbohydrates'] / df['Calories'] * 100
df.loc[:, ['Category', 'Item', 'Calories', 'Carbohydrates', 'Carbohydrates Percentage']]
```

## Agregando Columnas con Insert Method
Insertar una columna en una posición específica.

```python
# Insertar una columna en la posición 4
df.insert(4, 'Protein Percentage', df['Protein'] / df['Calories'] * 100, False)
df.head()
```

## Eliminando Columnas
Usar drop() para eliminar columnas.

```python
# Eliminar una columna
df.drop(columns='Protein Percentage', inplace=True)
df.head()
```

## Agregando Filas con Concat
Combinar dos DataFrames usando concat().

```python
# Crear primer DataFrame
student1 = {
    'Id': ['cs-01', 'mth-02', 'eng-03', 'phy-04', 'mth-05', 'cs-06'],
    'Name': ['Fatima', 'Noureen', 'Salman', 'Sana', 'Jack', 'Ali'],
    'subject': ['computer science', 'math', 'english', 'physics', 'statistics', 'programming'],
    'marks': [45, 90, 60, 37, 49, 50]
}
data1 = pd.DataFrame(student1)
data1
```

```python
# Crear segundo DataFrame
student2 = {
    'Id': ['cs-08', 'mth-07'],
    'Name': ['Adam', 'Steve'],
    'subject': ['computer science', 'math'],
    'marks': [45, 90]
}
data2 = pd.DataFrame(student2)
data2
```

```python
# Concatenar por filas (axis=0) manteniendo índices originales
pd.concat([data1, data2], ignore_index=False)
```

```python
# Concatenar por filas (axis=0) reseteando índices
pd.concat([data1, data2], ignore_index=True)
```

```python
# Concatenar por columnas (axis=1)
pd.concat([data1, data2], axis=1)
```

## Eliminando Filas
Usar drop() con el parámetro index para eliminar filas.

```python
# Concatenar y resetear índice
data1_combined = pd.concat([data1, data2], axis=0, ignore_index=True)
data1_combined
```

```python
# Eliminar filas específicas
data1_combined.drop(index=[0, 7])
```

## Conclusión
Agregar y eliminar filas y columnas es fundamental para la preparación y limpieza de datos. Estas operaciones son esenciales para preparar datasets para análisis.
