## Data Cleaning
En este notebook exploraremos técnicas adicionales de limpieza de datos. Aprenderemos a cambiar tipos de datos, interpolar y eliminar duplicados.

### Plan de aprendizaje:

```python
import pandas as pd
import numpy as np
```

## Casting: Conversión de Tipos de Datos
Cambiar el tipo de dato de una columna usando astype().

```python
# Crear un DataFrame de ejemplo
data = {'YearsCode': ['1', '2', '3', 'Less than 1 year', '5', 'More than 50 years', '10']}
df = pd.DataFrame(data)
df
```

```python
# Intentar convertir a float directamente (generará error)
# df['YearsCode'].astype(float)  # Esto causará error

# Primero, reemplazar valores textuales con numéricos
df['YearsCode'].replace({'Less than 1 year': 0, 'More than 50 years': 55}, inplace=True)
df
```

```python
# Ahora convertir a float
df['YearsCode'] = df['YearsCode'].astype(float)
df['YearsCode'].mean()
```

## Métodos de Interpolación: ffill y bfill
ffill: forward fill (propagar valores hacia adelante)
bfill: backward fill (propagar valores hacia atrás)

```python
# Crear un DataFrame con valores faltantes
data = pd.DataFrame(np.random.randn(7, 3))
data.iloc[:3, 1] = np.nan
data.iloc[:2, 2] = np.nan
data.iloc[4:6, 2] = np.nan
data
```

```python
# Forward fill: propagar el último valor conocido hacia adelante
data.fillna(method='ffill')
```

```python
# Backward fill: propagar el siguiente valor conocido hacia atrás
data.fillna(method='bfill')
```

```python
# ffill con límite: solo llenar hasta N valores
data.fillna(method='ffill', limit=1)
```

## Eliminando Duplicados
Usar duplicated() para detectar y drop_duplicates() para eliminar.

```python
# Crear un DataFrame con duplicados
data1 = pd.DataFrame({'col1': ['one', 'two'] * 4 + ['two'], 
                      'col2': [1, 1, 2, 3, 3, 4, 4, 5, 5]})
data1
```

```python
# Detectar duplicados (retorna valores booleanos)
data1.duplicated()
```

```python
# Contar el número de duplicados
data1.duplicated().sum()
```

```python
# Eliminar duplicados
data1_clean = data1.drop_duplicates()
data1_clean
```

## Conclusión
La limpieza de datos es un proceso continuo y esencial. Un dataset bien limpio es fundamental para análisis precisos y modelos de ML de alta calidad.
