# Advanced Indexing in Pandas

Learn to work with hierarchical multi-level indexes in Pandas for complex data structures.

```python
import pandas as pd
import numpy as np
```

## Series with MultiIndex

Create a Series with hierarchical two-level indexes.

```python
# Create a Series with MultiIndex
arrays = [['Team1', 'Team1', 'Team1', 'Team2', 'Team2', 'Team2', 'Team3', 'Team3', 'Team3'],
          ['player_1', 'player_2', 'player_3', 'player_1', 'player_2', 'player_3', 'player_1', 'player_2', 'player_3']]
index = pd.MultiIndex.from_arrays(arrays)
data = pd.Series([56, 78, 99, 56, 88, 59, 100, 57, 91], index=index, name="Player's Score")
data
```

```python
# View the MultiIndex of the Series
data.index
```

```python
# Select a range from level 0
data['Team1':'Team2']
```

## DataFrame with MultiIndex

Use set_index() to convert columns into MultiIndex.

```python
# Create a sample DataFrame
df = pd.DataFrame({
    'city': ['New York', 'New York', 'Los Angeles', 'Los Angeles', 'Chicago', 'Chicago'],
    'rank': [1, 2, 1, 2, 1, 2],
    'population': [8000000, 7000000, 4000000, 3500000, 2700000, 2300000],
    'area': [302, 302, 469, 469, 227, 227]
})
df
```

```python
# Set city and rank as MultiIndex
df.set_index(['city', 'rank'], inplace=True)
df
```

## Accessing MultiIndex Data

Access data using hierarchical index levels.

```python
# View the MultiIndex
df.index
```

```python
# Select a specific record using MultiIndex
df.loc['New York', 1]
```

## Reshaping: Stack and Unstack

Transform hierarchical index levels and columns.

```python
# Unstack: pivot innermost index level to columns
df.unstack()
```

```python
# Stack: pivot columns to index
df.stack()
```

## Sorting by Index Level

Sort data using the level parameter to specify which index level to sort by.

```python
# Sort by level 0 (city)
df.sort_index(level=0)
```

```python
# Sort by level 1 (rank)
df.sort_index(level=1)
```

## Conclusion

Advanced indexing with MultiIndex allows efficient work with hierarchical data structures. It's powerful for multidimensional analysis.
